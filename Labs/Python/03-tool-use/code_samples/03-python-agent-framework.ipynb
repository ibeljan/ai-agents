### 🔄 Expanding Tool Use: Step-by-Step

This section extends the basic travel agent into a richer multi-tool architecture and clarifies good vs bad tool design.

#### Incremental Extension Steps
1. Define clear purpose & contract (name, inputs, outputs, error modes)
2. Add thorough docstring (what it does, params, return schema, edge cases)
3. Use type hints to help validation & agent reasoning
4. Validate inputs early; fail fast with helpful errors
5. Prefer small pure functions; isolate side-effects (API calls, file I/O)
6. Return structured data (dict / pydantic model) when downstream parsing needed
7. Register tools together; keep list curated (avoid duplicates & overlap)
8. Log / annotate errors so agent can gracefully recover

#### Adding Multiple Tools
We'll add: weather lookup (stub), cost estimator, and local activity recommender. These complement the existing random destination generator.

#### Tool Contracts (Good Examples)
- get_weather_info(location: str) -> dict  | Shape: {"location": str, "forecast": str, "temp_c": float}
- calculate_travel_cost(destination: str, days: int, style: Literal['budget','standard','luxury']) -> dict
- recommend_local_activities(destination: str, interests: list[str], max_results: int = 5) -> list[dict]

#### Bad / Overlapping Examples
- random_place() and get_random_destination(): identical behavior, ambiguous naming
- getData(): vague purpose, no typing, mixes weather + cost + activities

Below we implement good tools, then show (commented) bad ones for comparison.

````python
# Good tool declarations (well-typed, single responsibility, validated)
from typing import Literal, List, Dict, Any

# ✅ Weather lookup tool (stubbed)
def get_weather_info(location: str) -> Dict[str, Any]:
    """
    Provides simple (stubbed) weather info for a given location.
    Args:
        location: City or destination name
    Returns:
        dict with keys: location(str), forecast(str), temp_c(float)
    Edge Cases:
        - Empty location -> ValueError
    """
    if not location or not location.strip():
        raise ValueError("location cannot be empty")
    # In real implementation: call external weather API here.
    # Stub logic: deterministic pseudo forecast based on hash.
    base = sum(ord(c) for c in location) % 3
    forecast_map = {0: "sunny", 1: "cloudy", 2: "rain showers"}
    forecast = forecast_map[base]
    # pseudo temperature
    temp_c = 18.0 + base * 4.5
    return {"location": location, "forecast": forecast, "temp_c": temp_c}

# ✅ Travel cost estimator
def calculate_travel_cost(destination: str, days: int, style: Literal['budget','standard','luxury']='standard') -> Dict[str, Any]:
    """
    Estimate travel cost.
    Args:
        destination: Destination name (non-empty)
        days: Number of days (>0)
        style: Travel style affecting daily rate
    Returns:
        dict with breakdown: {destination, days, style, daily_rate, total_estimate}
    """
    if not destination or not destination.strip():
        raise ValueError("destination cannot be empty")
    if days <= 0:
        raise ValueError("days must be > 0")
    rate_map = {'budget': 120, 'standard': 220, 'luxury': 480}
    daily_rate = rate_map[style]
    total = daily_rate * days
    return {
        "destination": destination,
        "days": days,
        "style": style,
        "daily_rate": daily_rate,
        "total_estimate": total
    }

# ✅ Local activities recommender
def recommend_local_activities(destination: str, interests: List[str], max_results: int = 5) -> List[Dict[str, Any]]:
    """
    Recommend activities matching interests.
    Args:
        destination: Target destination
        interests: List of interest keywords
        max_results: Limit returned activities
    Returns:
        list of activity dicts: [{name, category, destination}]
    Notes:
        - Deduplicates categories
        - Validates inputs
    """
    if not destination or not destination.strip():
        raise ValueError("destination cannot be empty")
    if not interests:
        raise ValueError("interests must contain at least one item")
    catalog = {
        'outdoors': ["Guided park walk", "Scenic bike tour", "Mountain lookout trek"],
        'food': ["Street food crawl", "Regional tasting", "Cooking workshop"],
        'culture': ["Museum tour", "Historic district walk", "Local gallery visit"],
        'relaxation': ["Thermal spa", "Beach sunset lounge", "Mindfulness retreat"],
    }
    # Map interests to categories (simple heuristic)
    selected_categories = []
    for interest in interests:
        interest_lower = interest.lower()
        if any(k in interest_lower for k in ["hike","bike","outdoor","nature"]):
            selected_categories.append('outdoors')
        elif any(k in interest_lower for k in ["food","cuisine","eat","restaurant"]):
            selected_categories.append('food')
        elif any(k in interest_lower for k in ["art","museum","history","culture"]):
            selected_categories.append('culture')
        elif any(k in interest_lower for k in ["relax","spa","meditate","zen"]):
            selected_categories.append('relaxation')
    # fallback
    if not selected_categories:
        selected_categories.append('culture')
    # Deduplicate while preserving order
    seen = set()
    categories = []
    for c in selected_categories:
        if c not in seen:
            categories.append(c); seen.add(c)
    activities: List[Dict[str, Any]] = []
    for cat in categories:
        for item in catalog.get(cat, []):
            activities.append({"name": item, "category": cat, "destination": destination})
            if len(activities) >= max_results:
                return activities
    return activities

print("✅ Tools defined: get_weather_info, calculate_travel_cost, recommend_local_activities")
````
