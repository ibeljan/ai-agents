# 🔷 AI Travel Agent with Microsoft Agent Framework (C#)

## 📋 Scenario Overview

This notebook demonstrates how to build an intelligent travel planning agent using the Microsoft Agent Framework for C#. The agent leverages Azure AI Foundry to automatically generate personalized day-trip itineraries for random destinations worldwide.

**Key Features:**
- 🎲 **Smart Destination Selection**: Custom tool function for random destination picking
- 🗺️ **Detailed Itinerary Generation**: AI-powered travel planning with local recommendations
- 🔄 **Async Processing**: Uses async/await for efficient API communication
- 🛠️ **Tool Integration**: Demonstrates function calling capabilities in AI agents

## 🏗️ Technical Implementation

### Core Components
- **Agent Framework**: C# implementation of Microsoft's agent orchestration system
- **Azure AI Foundry**: Access to state-of-the-art language models via Azure's AI platform
- **Azure Authentication**: Uses DefaultAzureCredential for secure Azure access
- **Environment Management**: Secure credential handling with .NET configuration

### Architecture Flow
```
User Request → ChatAgent → Azure AI Foundry API ↔ GetRandomDestination()
                     ↓
              Travel Itinerary Response
```

### Key Classes & Methods
- `ChatAgent`: Main conversational agent orchestrator
- `AzureAIAgentClient`: Azure AI Foundry API client wrapper
- `DefaultAzureCredential`: Azure authentication provider
- `GetRandomDestination()`: Custom tool function for destination selection


1. **Azure CLI**: Run `az login` to authenticate
2. **Or use other Azure credential methods** (Managed Identity, Service Principal, etc.)
3. **Azure AI Foundry Access**: Ensure your account has access to the AI Foundry resource

**Azure AI Foundry Setup:**
1. Create an Azure AI Foundry resource in Azure portal
2. Get the endpoint URL from the resource
3. Ensure your Azure account has appropriate permissions
4. Configure environment variables as shown above

````polyglot-notebook
// 📦 Import Required Namespaces
// Standard library imports for system operations
using System;
using System.Threading.Tasks;
using System.Collections.Generic;

// Azure authentication for AI Foundry
using Azure.Identity;
````

````polyglot-notebook
// 🤖 Import Microsoft Agent Framework Components
// ChatAgent: The main agent class for conversational AI
// AzureAIAgentClient: Client for connecting to Azure AI Foundry
using Azure.AI.Projects;
using Azure.AI.OpenAI;
using Azure.Identity;
using Microsoft.Agents.AI;
using OpenAI;
````

````polyglot-notebook
// 🔧 Load Environment Variables
// Get configuration from environment variables
// Required variables: AZURE_AI_FOUNDRY_ENDPOINT, AZURE_AI_FOUNDRY_MODEL

var modelDeploymentName = Environment.GetEnvironmentVariable("AZURE_AI_FOUNDRY_MODEL");
var projectEndpoint = Environment.GetEnvironmentVariable("AZURE_AI_FOUNDRY_ENDPOINT");

// Validate that environment variables are set
if (string.IsNullOrEmpty(projectEndpoint) || string.IsNullOrEmpty(modelDeploymentName))
{
    Console.ForegroundColor = ConsoleColor.Red;
    Console.WriteLine("❌ ERROR: Environment variables not set!");
    Console.ResetColor();
    Console.WriteLine("\nPlease set the following environment variables:");
    Console.WriteLine("  AZURE_AI_FOUNDRY_ENDPOINT=https://your-resource.openai.azure.com");
    Console.WriteLine("  AZURE_AI_FOUNDRY_MODEL=gpt-4o-mini");
    Console.WriteLine("\nTo set them in PowerShell:");
    Console.WriteLine("  $env:AZURE_AI_FOUNDRY_ENDPOINT='https://your-resource.openai.azure.com'");
    Console.WriteLine("  $env:AZURE_AI_FOUNDRY_MODEL='gpt-4o-mini'");
    throw new InvalidOperationException("Required environment variables are not set.");
}

Console.WriteLine($"✅ Model: {modelDeploymentName}");
Console.WriteLine($"✅ Endpoint: {projectEndpoint}");
````

````polyglot-notebook
// 🎲 Tool Function: Random Destination Generator
// This function will be available to the agent as a tool
// The agent can call this function to get random vacation destinations

string GetRandomDestination()
{
    // List of popular vacation destinations around the world
    var destinations = new List<string>
    {
        "Barcelona, Spain",
        "Paris, France",
        "Berlin, Germany",
        "Tokyo, Japan",
        "Sydney, Australia",
        "New York, USA",
        "Cairo, Egypt",
        "Cape Town, South Africa",
        "Rio de Janeiro, Brazil",
        "Bali, Indonesia"
    };
    
    // Return a random destination from the list
    var random = new Random();
    return destinations[random.Next(0, destinations.Count)];
}

// Test the function
Console.WriteLine($"Random destination: {GetRandomDestination()}");
````

````polyglot-notebook
// This client connects to Azure AI Foundry API
// Uses the environment variables loaded earlier for configuration

var agentChatClient = new AzureOpenAIClient(
    new Uri(projectEndpoint),
    new AzureCliCredential())
        .GetChatClient(modelDeploymentName);

Console.WriteLine("✅ Azure AI Agent Client created successfully");
````

````polyglot-notebook
// 🤖 Create the Travel Planning Agent
// This creates a conversational AI agent with specific capabilities:
// - chatClient: The AI model client for generating responses
// - instructions: System prompt that defines the agent's personality and role
// - tools: List of functions the agent can call to perform actions

var agent = new ChatAgent(
    chatClient: agentChatClient,
    instructions: "You are a helpful AI Agent that can help plan vacations for customers at random destinations.",
    tools: new[] { GetRandomDestination }
);

Console.WriteLine("✅ Travel Planning Agent created successfully");
````

````polyglot-notebook
// 🚀 Run the Agent
// Send a message to the agent and get a response
// The agent will use its tools (GetRandomDestination) if needed

var response = await agent.RunAsync("Plan me a day trip");

Console.WriteLine("✅ Agent response received");
````

````polyglot-notebook
// 📖 Extract and Display the Travel Plan
// Get the last message from the conversation (agent's response)
var lastMessage = response.Messages[response.Messages.Count - 1];

// Extract the text content from the message
var textContent = lastMessage.Contents[0].Text;

// Display the formatted travel plan
Console.WriteLine("🏖️ Travel plan:");
Console.WriteLine(textContent);
````
