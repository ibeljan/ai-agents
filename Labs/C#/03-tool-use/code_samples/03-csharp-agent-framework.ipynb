````python
// 📦 Import core dependencies for Agent Framework and tool integration
// This sets up the essential libraries for building intelligent agents with tool capabilities

using System;
using System.Threading.Tasks;
using System.Collections.Generic;
using System.Linq;

// Azure authentication
using Azure.Identity;
````

````python
// These are the core components for building tool-enabled agents
using Microsoft.AgentFramework;           // Main agent class
using Microsoft.AgentFramework.Azure;    // Azure AI integration
````

````python
// 🔑 Environment variables verification
// Ensure all required Azure AI Foundry credentials are properly configured

var projectEndpoint = Environment.GetEnvironmentVariable("AZURE_AI_FOUNDRY_PROJECT_ENDPOINT");
var modelDeploymentName = Environment.GetEnvironmentVariable("AZURE_AI_FOUNDRY_MODEL");

Console.WriteLine($"AZURE_AI_FOUNDRY_PROJECT_ENDPOINT: {projectEndpoint}");
Console.WriteLine($"AZURE_AI_FOUNDRY_MODEL: {modelDeploymentName}");
````

````python
// 🔗 Create Azure AI Agent Client
// Initialize the agent client with Azure credentials and endpoint configuration

var credential = new AzureCliCredential();

var agentChatClient = new AzureAIAgentClient(
    credential: credential,
    modelDeploymentName: modelDeploymentName,
    projectEndpoint: new Uri(projectEndpoint)
);

Console.WriteLine("✅ Azure AI Agent Client created successfully");
````

````python
// 🛠️ Define travel planning tools for agent integration
// These functions provide specific capabilities that the agent can invoke dynamically

string GetRandomDestination()
{
    /*
    🎲 Random destination generator tool
    Returns a randomly selected travel destination from curated list
    Useful when customers need inspiration for their next vacation
    */
    var destinations = new List<string>
    {
        "Paris, France",
        "Tokyo, Japan",
        "New York City, USA",
        "London, England",
        "Rome, Italy",
        "Sydney, Australia",
        "Dubai, UAE",
        "Barcelona, Spain",
        "Bangkok, Thailand",
        "Amsterdam, Netherlands",
        "Istanbul, Turkey",
        "Prague, Czech Republic",
        "Santorini, Greece",
        "Reykjavik, Iceland",
        "Marrakech, Morocco",
        "Cape Town, South Africa",
        "Rio de Janeiro, Brazil",
        "Bali, Indonesia"
    };
    
    // 🎯 Return random selection from the curated destination list
    var random = new Random();
    return destinations[random.Next(0, destinations.Count)];
}

// Test the tool
Console.WriteLine($"Sample destination: {GetRandomDestination()}");
````

````python
// 🤖 Configure travel agent identity and behavioral instructions
// Define the agent's personality, capabilities, and operational guidelines

const string AGENT_NAME = "TravelAgent";

const string AGENT_INSTRUCTIONS = "You are a helpful AI Agent that can help plan vacations for customers at random destinations";

Console.WriteLine($"Agent Name: {AGENT_NAME}");
Console.WriteLine($"Instructions: {AGENT_INSTRUCTIONS}");
````

````python
// 🏗️ Create the Travel Agent with Tool Integration
// Demonstrates the Agent Factory Pattern with tool registration

var agent = new ChatAgent(
    name: AGENT_NAME,
    chatClient: agentChatClient,
    instructions: AGENT_INSTRUCTIONS,
    tools: new[] { GetRandomDestination }
);

Console.WriteLine($"✅ Agent '{AGENT_NAME}' created with 1 tool");
````

````python
// 🧵 Create Conversation Thread
// Initialize a new thread for maintaining conversation context

var thread = agent.GetNewThread();

Console.WriteLine($"✅ Thread created: {thread.Id}");
````

````python
// 🚀 Execute Agent with Tool Usage
// The agent will automatically invoke the GetRandomDestination tool as needed

var response1 = await agent.RunAsync("Plan me a day trip", thread: thread);

Console.WriteLine("✅ Agent response received");
````

````python
// 📖 Display Travel Plan
// Extract and present the agent's tool-enhanced response

var lastMessage = response1.Messages[response1.Messages.Count - 1];
var textContent = lastMessage.Contents[0].Text;

Console.WriteLine("Travel plan:");
Console.WriteLine(textContent);
````
