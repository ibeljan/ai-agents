````python
// 📦 Import Standard Libraries
using System;
using System.Collections.Generic;
using System.Text.Json;
using System.Text.Json.Serialization;
````

````python
// 🤖 Import Microsoft Agent Framework Components
using Azure.Identity;
using Microsoft.AgentFramework;
using Microsoft.AgentFramework.Schema;
````

````python
// 🔧 Load Environment Variables
var modelDeploymentName = Environment.GetEnvironmentVariable("AZURE_AI_FOUNDRY_MODEL");
var projectEndpoint = Environment.GetEnvironmentVariable("AZURE_AI_FOUNDRY_PROJECT_ENDPOINT");

if (string.IsNullOrEmpty(modelDeploymentName) || string.IsNullOrEmpty(projectEndpoint))
{
    throw new InvalidOperationException("Required environment variables are not set. Please check your .env file.");
}
````

````python
// 🎯 Agent Configuration
const string AGENT_NAME = "TravelAgent";

// 📋 System Instructions for Planning Agent
const string AGENT_INSTRUCTIONS = @"You are a planner agent.
Your job is to decide which agents to run based on the user's request.
Below are the available agents specialized in different tasks:
- FlightBooking: For booking flights and providing flight information
- HotelBooking: For booking hotels and providing hotel information
- CarRental: For booking cars and providing car rental information
- ActivitiesBooking: For booking activities and providing activity information
";
````

````python
// 📝 Define Data Models for Structured Output

/// <summary>
/// Represents a single subtask in the travel planning process
/// </summary>
public class SubTask
{
    /// <summary>
    /// The specific agent assigned to handle this subtask
    /// </summary>
    [JsonPropertyName("assigned_agent")]
    public string AssignedAgent { get; set; } = string.Empty;

    /// <summary>
    /// Detailed description of what needs to be done for this subtask
    /// </summary>
    [JsonPropertyName("task_details")]
    public string TaskDetails { get; set; } = string.Empty;
}

/// <summary>
/// Represents the complete travel plan with main task and decomposed subtasks
/// </summary>
public class TravelPlan
{
    /// <summary>
    /// The overall travel request from the user
    /// </summary>
    [JsonPropertyName("main_task")]
    public string MainTask { get; set; } = string.Empty;

    /// <summary>
    /// List of subtasks broken down from the main task, each assigned to a specialized agent
    /// </summary>
    [JsonPropertyName("subtasks")]
    public List<SubTask> SubTasks { get; set; } = new List<SubTask>();
}
````

````python
// 🌐 Initialize Azure AI Agent Client
var credential = new AzureCliCredential();

var agentChatClient = new AzureAIAgentClient(
    credential: credential,
    modelDeploymentName: modelDeploymentName,
    projectEndpoint: new Uri(projectEndpoint)
);
````

````python
// 🤖 Create Planning Agent with Structured Output
var agent = new ChatAgent(
    name: AGENT_NAME,
    chatClient: agentChatClient,
    instructions: AGENT_INSTRUCTIONS
);
````

````python
// 💬 Create User Message
var userMessage = "Create a travel plan for a family of 4, with 2 kids, from Singapore to Melbourne";
Console.WriteLine($"User Request: {userMessage}");
````

````python
// 🚀 Invoke Agent with Structured Output Format
var chatOptions = new ChatOptions
{
    ResponseFormat = ChatResponseFormat.CreateJsonSchemaFormat<TravelPlan>(
        name: "TravelPlan",
        description: "A structured travel plan with main task and subtasks",
        strictSchemaEnabled: true
    )
};

var response = await agent.InvokeAsync(userMessage, chatOptions);
var responseText = response.GetValue<string>();

Console.WriteLine($"\nAgent Response:\n{responseText}");
````

````python
// 📊 Parse and Display Structured Travel Plan
var travelPlan = JsonSerializer.Deserialize<TravelPlan>(responseText);

if (travelPlan != null)
{
    Console.WriteLine($"\n🎯 Main Task: {travelPlan.MainTask}");
    Console.WriteLine($"\n📋 Subtasks ({travelPlan.SubTasks.Count}):");
    
    for (int i = 0; i < travelPlan.SubTasks.Count; i++)
    {
        var subtask = travelPlan.SubTasks[i];
        Console.WriteLine($"\n{i + 1}. Agent: {subtask.AssignedAgent}");
        Console.WriteLine($"   Task: {subtask.TaskDetails}");
    }
}
````
